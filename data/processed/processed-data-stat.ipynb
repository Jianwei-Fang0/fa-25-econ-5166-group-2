## 1. NTU YouBike 交易資料
- Unit of observation: 單次 YouBike 租借交易紀錄
- Sample period: 2024年3月 - 2025年1月
- Definition for each variable: 
    - borrow_datetime: 借車時間（日期時間格式）
    - borrow_site: 借車站點名稱
    - return_datetime: 還車時間（日期時間格式）
    - return_site: 還車站點名稱
    - borrowing_time: 租借時長（時間格式，HH:MM:SS）
    - type: 車輛類型（一般車/電輔車）
    - date: 交易日期

## 2. 站點流量資料
- Unit of observation: 站點 × 日期 × 小時的流量紀錄
- Sample period: 2024年3月 - 2025年1月
- Definition for each variable:
    - station: 站點名稱
    - date: 日期
    - hour: 小時（0-23）
    - borrow_count: 該時段借車次數
    - return_count: 該時段還車次數

## 3. 站點資訊
- Unit of observation: 單一站點的基本資訊
- Definition for each variable:
    - sno: 站點編號
    - sna: 站點名稱
    - sarea: 區域
    - latitude: 緯度
    - longitude: 經度
    - ar: 站點地址
    - sareaen: 區域（英文）
    - aren: 地址（英文）
    - act: 啟用狀態（0=未啟用，1=啟用）
    - total: 總車位數

## 4. 站點間路線資訊
- Unit of observation: 站點間的路線資訊
- Definition for each variable:
    - origin: 起點站點名稱
    - destination: 終點站點名稱
    - distance_km: 距離（公里）
    - duration_min: 預估時間（分鐘）
    - origin_sno: 起點站點編號
    - destination_sno: 終點站點編號

```python
import pandas as pd
import glob
import os
```

## 資料設定

```python
# 選擇要分析的資料檔案（可選擇單一月份或合併所有月份）
# 單一月份範例：
input_data_file = "./ntu_youbike_data/ntu_youbike_data_202501.csv"

# 或合併所有月份資料（取消註解以使用）：
# ntu_files = glob.glob("./ntu_youbike_data/*.csv")
# ntu_files.sort()
# input_data_file = ntu_files  # 將讀取多個檔案
```

```python
# 讀取資料
if isinstance(input_data_file, list):
    # 如果 input_data_file 是列表，合併所有檔案
    df_list = []
    for file in input_data_file:
        df_temp = pd.read_csv(file)
        df_list.append(df_temp)
    df = pd.concat(df_list, ignore_index=True)
else:
    # 讀取單一檔案
    df = pd.read_csv(input_data_file)

print(f"資料形狀: {df.shape}")
print(f"欄位: {list(df.columns)}")
```

### Assertions

```python
# Write assertions here—conditions that should always hold in the data.
# Use them to demonstrate that you have carefully checked the data’s correctness.

```

```python
# 基本資料完整性檢查
assert df.shape[0] > 0, "資料不應為空"
assert 'borrow_datetime' in df.columns, "應包含借車時間欄位"
assert 'return_datetime' in df.columns, "應包含還車時間欄位"

# 檢查缺失值
print("各欄位缺失值數量:")
print(df.isnull().sum())

# 檢查資料類型
print("\n各欄位資料類型:")
print(df.dtypes)

# 檢查時間邏輯：還車時間應該晚於借車時間（轉換為datetime後檢查）
if 'borrow_datetime' in df.columns and 'return_datetime' in df.columns:
    df['borrow_dt'] = pd.to_datetime(df['borrow_datetime'])
    df['return_dt'] = pd.to_datetime(df['return_datetime'])
    invalid_time = (df['return_dt'] < df['borrow_dt']).sum()
    print(f"\n還車時間早於借車時間的記錄數: {invalid_time}")
    assert invalid_time == 0, "所有還車時間應晚於借車時間"

# 檢查車輛類型
if 'type' in df.columns:
    print(f"\n車輛類型分布:")
    print(df['type'].value_counts())
    
# 檢查日期範圍
if 'date' in df.columns:
    df['date_parsed'] = pd.to_datetime(df['date'])
    print(f"\n資料日期範圍: {df['date_parsed'].min()} 至 {df['date_parsed'].max()}")
```

## 站點流量資料分析


```python
# 選擇要分析的站點流量資料檔案
# 單一月份範例：
station_flow_file = "./station_flow/station_flow_202501.csv"

# 或合併所有月份站點流量資料（取消註解以使用）：
# flow_files = glob.glob("./station_flow/*.csv")
# flow_files.sort()
# station_flow_file = flow_files  # 將讀取多個檔案
```

```python
# 讀取站點流量資料
if isinstance(station_flow_file, list):
    # 如果 station_flow_file 是列表，合併所有檔案
    flow_df_list = []
    for file in station_flow_file:
        df_temp = pd.read_csv(file)
        flow_df_list.append(df_temp)
    df_flow = pd.concat(flow_df_list, ignore_index=True)
else:
    # 讀取單一檔案
    df_flow = pd.read_csv(station_flow_file)

print(f"站點流量資料形狀: {df_flow.shape}")
print(f"欄位: {list(df_flow.columns)}")
```

### 站點流量資料樣本


```python
df_flow.head(10)
```

### 站點流量統計摘要


```python
# 數值欄位的統計摘要
if 'borrow_count' in df_flow.columns and 'return_count' in df_flow.columns:
    print("借車/還車數量統計:")
    print(df_flow[['borrow_count', 'return_count']].describe())
    print("\n總借車次數:", df_flow['borrow_count'].sum())
    print("總還車次數:", df_flow['return_count'].sum())
    
# 站點數量統計
if 'station' in df_flow.columns:
    print(f"\n唯一站點數量: {df_flow['station'].nunique()}")
    
# 日期範圍
if 'date' in df_flow.columns:
    df_flow['date_parsed'] = pd.to_datetime(df_flow['date'])
    print(f"日期範圍: {df_flow['date_parsed'].min()} 至 {df_flow['date_parsed'].max()}")
    
# 小時分布
if 'hour' in df_flow.columns:
    print(f"\n小時分布:")
    print(df_flow['hour'].value_counts().sort_index())
```

## 站點間路線資訊分析


```python
# 讀取站點間路線資訊
routes_df = pd.read_csv("./routes_with_sno_clean.csv")
print(f"路線資訊資料形狀: {routes_df.shape}")
print(f"欄位: {list(routes_df.columns)}")
```

### 路線資訊樣本


```python
routes_df.head(10)
```

### 路線資訊統計摘要


```python
# 路線總數
print(f"總路線數: {len(routes_df)}")

# 唯一起點和終點
if 'origin' in routes_df.columns and 'destination' in routes_df.columns:
    print(f"唯一起點數: {routes_df['origin'].nunique()}")
    print(f"唯一終點數: {routes_df['destination'].nunique()}")

# 距離和時間統計
if 'distance_km' in routes_df.columns:
    print(f"\n路線距離統計 (公里):")
    print(routes_df['distance_km'].describe())

if 'duration_min' in routes_df.columns:
    print(f"\n路線時間統計 (分鐘):")
    print(routes_df['duration_min'].describe())

# 最長/最短路線
if 'distance_km' in routes_df.columns:
    longest = routes_df.loc[routes_df['distance_km'].idxmax()]
    shortest = routes_df.loc[routes_df['distance_km'].idxmin()]
    print(f"\n最長路線: {longest['origin']} -> {longest['destination']} ({longest['distance_km']:.2f} 公里)")
    print(f"最短路線: {shortest['origin']} -> {shortest['destination']} ({shortest['distance_km']:.2f} 公里)")
```
