# Data Description
- Background: 台北市YouBike2.0的交易資料以及大安區的天氣資料，用於分析台大校內YouBike的使用情況及天氣對其的影響。
- date received: 2024/03/01
- Path to data file: /datafiles
- Unit of observation: Weather -- 8784
- Sample period: 2024/03~2024/12
- Known issues:
    - 原始資料中降水量資料在雨量小於0.5mm時只會用T表示
    - 
- Definition for each variable: 
    - sno -- 站點代號
    - sna -- 站點中文名稱
    - sarea -- 站點所在區域的中文名稱
    - sareaen -- 站點所在區域的英文名稱
    - ar -- 站點地址中文名稱
    - aren -- 站點地址英文名稱
    - laititude -- 站點的緯度
    - longtitude -- 站點的經度
    - act -- 站點的禁用狀態，1表示啟用
    - total -- 站點的車位數

    - origin -- 起始站點
    - destination -- 目標站點
    - distance_km -- 兩個站點之間的距離，以公里表示
    - duration_min -- 兩個站點之間移動所需的時間，以分鐘表示
    - origin_sno -- 起始站點代號
    - destination_sno -- 目標站點代號

    - station -- 站點地址
    - date -- 在此站點借還車的日期
    - hour -- 在此站點借還車的時間
    - borrow_count -- 站點在1個小時間的借車人數
    - return_count -- 佔點個1個小時間的還車人數

    - ObsDate -- 天氣資訊的觀測日期
    - ObsTime -- 天氣資訊的觀測時間
    - StnPres -- 大安區的氣壓，單位為hPa
    - SeaPres -- 海平面氣壓，單位為hPa
    - Temperature -- 大安區的氣溫，單位為℃
    - Td dew point -- 大安區的露點溫度，單位為℃
    - RH -- 大安區的相對溼度
    - WS -- 大安區測到的風速，單位為m/s
    - WD -- 大安區的測到的風向
    - WSGust -- 最大瞬間風速，單位為m/s
    - WDGust -- 最大瞬間風風向
    - Precp -- 大安區的降水量，單位為mm，若降水量小於0.5mm則以T表示
    - PrecpHour -- 降水時數
    - GloblRad -- 大安區的全天空日射量，單位為MJ/㎡
    - UVI -- 大安區的紫外線指數
